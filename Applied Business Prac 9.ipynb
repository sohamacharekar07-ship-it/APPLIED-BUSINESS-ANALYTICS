```python
import pandas as pd
import scipy.stats as stats

# Load dataset
df = pd.read_csv("supermarket_sales.csv")

# Show first 5 records
print(df.head())

# Use 'revenue' column for hypothesis testing
revenue = df['revenue']

# Check missing values
print("Missing values:", revenue.isnull().sum())

# Mean revenue
mean_revenue = revenue.mean()
print("Mean Revenue:", mean_revenue)

# Hypothesis setup
expected_mean = 1500
alpha = 0.05

# One-sample t-test
t_stat, p_value = stats.ttest_1samp(revenue, expected_mean)

print("T-statistic:", t_stat)
print("P-value:", p_value)

# Decision
if p_value < alpha:
    print("Reject Null Hypothesis")
    print("Conclusion: The average revenue is significantly different from ₹1500.")
else:
    print("Fail to Reject Null Hypothesis")
    print("Conclusion: The average revenue is not significantly different from ₹1500.")
```

```python
import numpy as np
from statsmodels.stats.proportion import proportions_ztest
import matplotlib.pyplot as plt

# A/B testing dataset
data = {
    'Group': ['A', 'B'],
    'Users': [1000, 1000],
    'Conversions': [120, 150]
}

# Display data
print(data)

# Conversion rates
conv_rate_A = data['Conversions'][0] / data['Users'][0]
conv_rate_B = data['Conversions'][1] / data['Users'][1]

print("Conversion Rate A:", conv_rate_A)
print("Conversion Rate B:", conv_rate_B)

# Hypothesis setup
alpha = 0.05

# Two-proportion z-test
successes = np.array([data['Conversions'][0], data['Conversions'][1]])
nobs = np.array([data['Users'][0], data['Users'][1]])

z_stat, p_value = proportions_ztest(successes, nobs, alternative='two-sided')

print("Z-statistic:", z_stat)
print("P-value:", p_value)

# Decision
if p_value < alpha:
    print("Reject Null Hypothesis")
    print("Conclusion: Conversion rates between Group A and Group B are significantly different.")
else:
    print("Fail to Reject Null Hypothesis")
    print("Conclusion: No significant difference in conversion rates between Group A and Group B.")

# Bar chart
plt.bar(data['Group'], [conv_rate_A, conv_rate_B], color=['blue', 'green'])
plt.ylabel("Conversion Rate")
plt.title("A/B Testing Conversion Rates")
plt.show()
```
