a. Simulate a probability experiment (e.g., rolling dice) using programming
and calculate the probabilities of different outcomes.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
print("========== DICE ROLL EXPERIMENT ==========")
```

```python
# 1. Simulate 1,000 dice rolls

rolls = np.random.randint(1, 7, 1000)

print("\nFirst 20 Dice Outcomes:")
print(rolls[:20])
```

```python
# 2. Calculate experimental probability
frequency = pd.Series(rolls).value_counts().sort_index()
experimental_probability = frequency / 1000
print("\nFrequency of Each Outcome:")
print(frequency)
print("\nExperimental Probability:")
for number in range(1, 7):
    print(f"P({number}) = {experimental_probability[number]:.4f}")
```

```python
# 3. Calculate theoretical probability
theoretical_probability = 1 / 6
print("\nTheoretical Probability:")
for number in range(1, 7):
    print(f"P({number}) = {theoretical_probability:.4f}")
```

```python
# 4. Probability of getting 1, 2, 3, 4, 5, 6
print("\nProbability of Individual Outcomes:")
for number in range(1, 7):
    print(f"Number {number}: {experimental_probability[number]:.4f}")
```

```python
# 5. Probability of getting an even number
even_probability = np.mean(np.isin(rolls, [2, 4, 6]))
print("\nProbability of Even Number:")
print(f"Experimental Probability = {even_probability:.4f}")
print(f"Theoretical Probability = {3/6:.4f}")
```

```python
# 6. Probability of getting an odd number
odd_probability = np.mean(np.isin(rolls, [1, 3, 5]))
print("\nProbability of Odd Number:")
print(f"Experimental Probability = {odd_probability:.4f}")
print(f"Theoretical Probability = {3/6:.4f}")
```

```python
# 7. Probability of getting a number greater than 4
greater_4 = np.mean(rolls > 4)
print("\nProbability of Number Greater Than 4:")
print(f"Experimental Probability = {greater_4:.4f}")
print(f"Theoretical Probability = {2/6:.4f}")
```

```python
# 8. Probability of getting a number less than 4
less_4 = np.mean(rolls < 4)
print("\nProbability of Number Less Than 4:")
print(f"Experimental Probability = {less_4:.4f}")
print(f"Theoretical Probability = {3/6:.4f}")
```

```python
# 9. Probability of getting a number greater than or equal to 4
greater_equal_4 = np.mean(rolls >= 4)
print("\nProbability of Number Greater Than or Equal To 4:")
print(f"Experimental Probability = {greater_equal_4:.4f}")
print(f"Theoretical Probability = {3/6:.4f}")
```

```python
# 10. Bar chart showing frequency of outcomes 1 to 6
plt.figure(figsize=(8, 5))
plt.bar(frequency.index, frequency.values)
plt.xlabel("Dice Outcome")
plt.ylabel("Frequency")
plt.title("Frequency of Dice Outcomes")
plt.xticks(range(1, 7))
plt.show()
```

```python
# 11. Bar chart showing experimental probability
plt.figure(figsize=(8, 5))
plt.bar(
    experimental_probability.index,
    experimental_probability.values
)
plt.xlabel("Dice Outcome")
plt.ylabel("Experimental Probability")
plt.title("Experimental Probability of Dice Outcomes")
plt.xticks(range(1, 7))
plt.show()
```

```python
# 12. Pie chart showing percentage distribution
plt.figure(figsize=(7, 7))
plt.pie(
    frequency.values,
    labels=frequency.index,
    autopct="%1.1f%%"
)
plt.title("Percentage Distribution of Dice Outcomes")
plt.show()
```

```python
# 13. Line chart comparing experimental and theoretical probability
plt.figure(figsize=(8, 5))
plt.plot(
    range(1, 7),
    experimental_probability.values,
    marker="o",
    label="Experimental Probability"
)
plt.plot(
    range(1, 7),
    [theoretical_probability] * 6,
    marker="o",
    label="Theoretical Probability"
)
plt.xlabel("Dice Outcome")
plt.ylabel("Probability")
plt.title("Experimental vs Theoretical Probability")
plt.xticks(range(1, 7))
plt.legend()
plt.grid()
plt.show()
```

```python
# 14. Chart showing frequency of Even vs Odd outcomes
even_count = np.sum(np.isin(rolls, [2, 4, 6]))
odd_count = np.sum(np.isin(rolls, [1, 3, 5]))
plt.figure(figsize=(7, 5))
plt.bar(
    ["Even", "Odd"],
    [even_count, odd_count]
)
plt.xlabel("Category")
plt.ylabel("Frequency")
plt.title("Frequency of Even vs Odd Outcomes")
plt.show()
```

```python
# 15. Experimental probability of Even, Odd, Greater than 4, Less than 4
category_probabilities = {
    "Even": even_probability,
    "Odd": odd_probability,
    "> 4": greater_4,
    "< 4": less_4
}
plt.figure(figsize=(9, 5))
plt.bar(
    category_probabilities.keys(),
    category_probabilities.values()
)
plt.xlabel("Category")
plt.ylabel("Experimental Probability")
plt.title("Experimental Probability of Dice Categories")
plt.show()
print("\n========== DICE EXPERIMENT COMPLETED ==========")
```

b. Generate random numbers from various probability distributions (normal,
uniform, exponential) and analyze their properties.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as pl
```

```python
np.random.seed(42)
print("========== PROBABILITY DISTRIBUTIONS ==========")
```

```python
# NORMAL DISTRIBUTION
normal_data = np.random.normal(
    loc=50,
    scale=10,
    size=1000
)
normal_series = pd.Series(normal_data)
```

```python
# 1. Generate 1,000 random numbers from normal distribution
print("\n========== NORMAL DISTRIBUTION ==========")
print("First 20 Normal Distribution Values:")
print(normal_data[:20])
```

```python
# 2. Mean, Median, Mode, Variance and Standard Deviation
mean_normal = normal_series.mean()
median_normal = normal_series.median()
mode_normal = normal_series.mode()[0]
variance_normal = normal_series.var()
std_normal = normal_series.std()
print(f"\nMean = {mean_normal:.4f}")
print(f"Median = {median_normal:.4f}")
print(f"Mode = {mode_normal:.4f}")
print(f"Variance = {variance_normal:.4f}")
print(f"Standard Deviation = {std_normal:.4f}")
```

```python
# 3. Statistics of Normal Distribution
minimum_normal = normal_series.min()
maximum_normal = normal_series.max()
print("\nNormal Distribution Statistics:")
print(f"Mean = {mean_normal:.4f}")
print(f"Median = {median_normal:.4f}")
print(f"Variance = {variance_normal:.4f}")
print(f"Standard Deviation = {std_normal:.4f}")
print(f"Minimum = {minimum_normal:.4f}")
print(f"Maximum = {maximum_normal:.4f}")
```

```python
# 4. Histogram of Normal Distribution
plt.figure(figsize=(8, 5))
plt.hist(normal_data, bins=30)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram of Normal Distribution")
plt.show()
```

```python
# 5. Box plot of Normal Distribution
plt.figure(figsize=(8, 5))
plt.boxplot(normal_data)
plt.ylabel("Value")
plt.title("Box Plot of Normal Distribution")
plt.show()
```

```python
# 6. Percentage of values between 40 and 60
between_40_60 = np.mean(
    (normal_data >= 40) &
    (normal_data <= 60)
) * 100
print(
    f"\nPercentage of values between 40 and 60 = "
    f"{between_40_60:.2f}%"
)
```

```python
# 7. Percentage of values between 30 and 70
between_30_70 = np.mean(
    (normal_data >= 30) &
    (normal_data <= 70)
) * 100
print(
    f"Percentage of values between 30 and 70 = "
    f"{between_30_70:.2f}%"
)
# UNIFORM DISTRIBUTION
uniform_data = np.random.uniform(
    low=0,
    high=100,
    size=1000
)
print("\n========== UNIFORM DISTRIBUTION ==========")
print("First 20 Uniform Distribution Values:")
print(uniform_data[:20])
print(f"\nMean = {np.mean(uniform_data):.4f}")
print(f"Median = {np.median(uniform_data):.4f}")
print(f"Minimum = {np.min(uniform_data):.4f}")
print(f"Maximum = {np.max(uniform_data):.4f}")
```

```python
# 9. Histogram of Uniform Distribution
plt.figure(figsize=(8, 5))
plt.hist(uniform_data, bins=20)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram of Uniform Distribution")
plt.show()
```

```python
# 10. Box plot of Uniform Distribution
plt.figure(figsize=(8, 5))
plt.boxplot(uniform_data)
plt.ylabel("Value")
plt.title("Box Plot of Uniform Distribution")
plt.show()
# EXPONENTIAL DISTRIBUTION
exponential_data = np.random.exponential(
    scale=2,
    size=1000
)
print("\n========== EXPONENTIAL DISTRIBUTION ==========")
print("First 20 Exponential Distribution Values:")
print(exponential_data[:20])
print(f"\nMean = {np.mean(exponential_data):.4f}")
print(f"Median = {np.median(exponential_data):.4f}")
print(f"Minimum = {np.min(exponential_data):.4f}")
print(f"Maximum = {np.max(exponential_data):.4f}")
```

```python
# 12. Histogram of Exponential Distribution
plt.figure(figsize=(8, 5))
plt.hist(exponential_data, bins=30)
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram of Exponential Distribution")
plt.show()
```

```python
# 13. Box plot of Exponential Distribution
plt.figure(figsize=(8, 5))
plt.boxplot(exponential_data)
plt.ylabel("Value")
plt.title("Box Plot of Exponential Distribution")
plt.show()
```

```python
# 14. Percentage of values greater than 2
greater_than_2 = np.mean(
    exponential_data > 2
) * 100
print(
    f"\nPercentage of values greater than 2 = "
    f"{greater_than_2:.2f}%"
)
```

```python
# 15. Percentage of values greater than 5
greater_than_5 = np.mean(
    exponential_data > 5
) * 100
print(
    f"Percentage of values greater than 5 = "
    f"{greater_than_5:.2f}%"
)
```

# C) Coin toss task

```python

import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

print("========== COIN TOSS EXPERIMENT ==========")

# 1. Simulate 1,000 coin tosses
tosses = np.random.choice(
    ["Heads", "Tails"],
    size=1000
)
heads = np.sum(tosses == "Heads")
tails = np.sum(tosses == "Tails")
experimental_heads = heads / 1000
experimental_tails = tails / 1000
print("\nTotal Tosses:", 1000)
print("Number of Heads:", heads)
print("Number of Tails:", tails)
print("\nExperimental Probability:")
print(f"Heads = {experimental_heads:.4f}")
print(f"Tails = {experimental_tails:.4f}")

# 2. Calculate theoretical probability
theoretical_heads = 0.5
theoretical_tails = 0.5
print("\nTheoretical Probability:")
print(f"Heads = {theoretical_heads:.4f}")
print(f"Tails = {theoretical_tails:.4f}")

# 3. Bar chart showing frequency of Heads and Tails
plt.figure(figsize=(7, 5))
plt.bar(
    ["Heads", "Tails"],
    [heads, tails]
)
plt.xlabel("Outcome")
plt.ylabel("Frequency")
plt.title("Frequency of Heads and Tails")
plt.show()

# 4. Pie chart showing percentage of Heads and Tails
plt.figure(figsize=(7, 7))
plt.pie(
    [heads, tails],
    labels=["Heads", "Tails"],
    autopct="%1.1f%%"
)
plt.title("Percentage Distribution of Heads and Tails")
plt.show()

# 5. Coin tossed 10 times and Heads occurs 8 times
heads_10 = 8
total_10 = 10
probability_heads_10 = heads_10 / total_10
print("\n========== 10 TOSS EXAMPLE ==========")
print("Total Tosses = 10")
print("Number of Heads = 8")
print(
    f"Experimental Probability of Heads = "
    f"{probability_heads_10:.2f}"
)

# 6. Simulate 10,000 coin tosses
tosses_10000 = np.random.choice(
    ["Heads", "Tails"],
    size=10000
)
heads_10000 = np.sum(tosses_10000 == "Heads")
tails_10000 = np.sum(tosses_10000 == "Tails")
experimental_heads_10000 = heads_10000 / 10000
experimental_tails_10000 = tails_10000 / 10000
difference_heads = abs(
    experimental_heads_10000 -
    theoretical_heads
)
difference_tails = abs(
    experimental_tails_10000 -
    theoretical_tails
)
print("\n========== 10,000 COIN TOSS ANALYSIS ==========")
print(f"Total Tosses = 10000")
print(f"Number of Heads = {heads_10000}")
print(f"Number of Tails = {tails_10000}")
print(
    f"Experimental Probability of Heads = "
    f"{experimental_heads_10000:.4f}"
)
print(
    f"Experimental Probability of Tails = "
    f"{experimental_tails_10000:.4f}"
)
print(
    f"Theoretical Probability of Heads = "
    f"{theoretical_heads:.4f}"
)
print(
    f"Theoretical Probability of Tails = "
    f"{theoretical_tails:.4f}"
)
print(
    f"Difference between Experimental and Theoretical "
    f"Probability of Heads = {difference_heads:.4f}"
)
print(
    f"Difference between Experimental and Theoretical "
    f"Probability of Tails = {difference_tails:.4f}"
)

# Bar chart for 10,000 tosses
plt.figure(figsize=(7, 5))
plt.bar(
    ["Heads", "Tails"],
    [heads_10000, tails_10000]
)
plt.xlabel("Outcome")
plt.ylabel("Frequency")
plt.title("Heads vs Tails - 10,000 Coin Tosses")
plt.show()

# Pie chart for 10,000 tosses
plt.figure(figsize=(7, 7))
plt.pie(
    [heads_10000, tails_10000],
    labels=["Heads", "Tails"],
    autopct="%1.2f%%"
)
plt.title("Percentage Distribution - 10,000 Coin Tosses")
plt.show()
print("\n========== COIN TOSS EXPERIMENT COMPLETED ==========")
```
