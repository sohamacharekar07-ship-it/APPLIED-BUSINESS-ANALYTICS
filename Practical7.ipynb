```python
# PART A - Survey and Sample Data Analysis

# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the Anti-Ragging Awareness CSV dataset
df = pd.read_csv("Anti_Ragging_Awareness_CEP_Dataset.csv")

print("Dataset Loaded Successfully!")
print("Total Records:", len(df))
print("Total Columns:", len(df.columns))

# Display first five records
display(df.head())

# Identify population
print("\nPopulation:")
print("All people in the target community who could participate in the Anti-Ragging Awareness survey.")

# Identify sample
print("\nSample:")
print("75 respondents included in the Anti-Ragging Awareness survey dataset.")

# Identify sampling technique
print("\nSampling Technique:")
print("Convenience Sampling")

# Identify numerical variables
numerical_variables = [
    "Seriousness Rating",
    "Information Useful",
    "Awareness Increased"
]

print("\nNumerical Variables:")
for variable in numerical_variables:
    print(variable)

# Identify categorical variables
categorical_variables = [
    "Name",
    "Age Group",
    "Gender",
    "Occupation",
    "Education Level",
    "Location",
    "Heard About Ragging",
    "Knows What Is Ragging",
    "Knows It Is Prohibited",
    "Knows Reporting Method",
    "Knows Helpline",
    "Witnessed/Experienced",
    "Awareness Programs Helpful",
    "Will Share Information",
    "Feedback"
]

print("\nCategorical Variables:")
for variable in categorical_variables:
    print(variable)

# Select numerical variable for analysis
selected_variable = "Awareness Increased"

# Remove missing values
data = df[selected_variable].dropna()

# Calculate sample size
sample_size = data.count()

# Calculate mean
sample_mean = data.mean()

# Calculate median
sample_median = data.median()

# Calculate standard deviation
sample_std = data.std()

print("\nSelected Variable:", selected_variable)
print("Sample Size:", sample_size)
print("Mean:", round(sample_mean, 2))
print("Median:", sample_median)
print("Standard Deviation:", round(sample_std, 2))

# Create histogram
plt.figure(figsize=(8, 5))

plt.hist(
    data,
    bins=np.arange(0.5, 6.5, 1),
    edgecolor="black"
)

plt.title("Awareness Increased Among Respondents")
plt.xlabel("Awareness Increased Rating")
plt.ylabel("Number of Respondents")
plt.xticks([1, 2, 3, 4, 5])
plt.grid(axis="y", alpha=0.3)

plt.show()

# Write two observations
print("\nObservations:")
print("1. Most respondents reported a moderate to high increase in awareness.")
print("2. The survey shows that the awareness activity helped improve understanding about anti-ragging.")
```

```python
# PART B - Sampling Distribution

# Select numerical variable
selected_variable = "Awareness Increased"

# Remove missing values
data = df[selected_variable].dropna()

# Select a random sample of 30 observations
random_sample_30 = data.sample(
    n=30,
    random_state=42
)

# Calculate sample mean
sample_30_mean = random_sample_30.mean()

print("\nRandom Sample of 30 Observations:")
print(random_sample_30.values)

print("\nMean of Selected Sample of 30:")
print(round(sample_30_mean, 2))

# Repeat random sampling 100 times
sample_means_30 = []

for i in range(100):
    sample = data.sample(
        n=30,
        replace=True
    )

    sample_mean_value = sample.mean()
    sample_means_30.append(sample_mean_value)

print("\n100 Random Samples Generated Successfully.")

# Store sample means in a DataFrame
sampling_distribution = pd.DataFrame({
    "Sample Number": range(1, 101),
    "Sample Mean": sample_means_30
})

# Display first 10 sample means
print("\nFirst 10 Sample Means:")
display(sampling_distribution.head(10))

# Calculate mean of all sample means
mean_of_sample_means = pd.Series(sample_means_30).mean()

print("\nMean of All Sample Means:")
print(round(mean_of_sample_means, 2))

# Calculate standard deviation of sample means
std_of_sample_means = pd.Series(sample_means_30).std()

print("\nStandard Deviation of Sample Means:")
print(round(std_of_sample_means, 2))

# Create histogram of sampling distribution
plt.figure(figsize=(8, 5))

plt.hist(
    sample_means_30,
    bins=10,
    edgecolor="black"
)

plt.axvline(
    mean_of_sample_means,
    linestyle="--",
    linewidth=2,
    label="Mean of Sample Means"
)

plt.title("Sampling Distribution of Sample Means (n = 30)")
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")
plt.legend()
plt.grid(axis="y", alpha=0.3)

plt.show()

# Observe the shape
print("\nShape of Sampling Distribution:")
print("The sampling distribution is approximately bell-shaped and concentrated around its mean.")

# Compare original mean and sampling distribution mean
original_mean = data.mean()

print("\nComparison of Means:")
print("Original Dataset Mean:", round(original_mean, 2))
print("Mean of Sampling Distribution:", round(mean_of_sample_means, 2))

difference = abs(
    original_mean - mean_of_sample_means
)

print("Difference:", round(difference, 2))

# Compare sampling distributions for different sample sizes
sample_sizes = [10, 30, 50]

sampling_results = {}

for size in sample_sizes:

    means = []

    # Generate 100 samples for each sample size
    for i in range(100):

        sample = data.sample(
            n=size,
            replace=True
        )

        means.append(sample.mean())

    sampling_results[size] = means

    means_series = pd.Series(means)

    print(
        "Sample Size:", size,
        "| Mean:", round(means_series.mean(), 2),
        "| Standard Deviation:", round(means_series.std(), 2)
    )

# Plot sampling distributions for different sample sizes
plt.figure(figsize=(9, 6))

for size in sample_sizes:

    plt.hist(
        sampling_results[size],
        bins=10,
        alpha=0.5,
        label="Sample Size " + str(size)
    )

plt.title("Sampling Distributions for Different Sample Sizes")
plt.xlabel("Sample Mean")
plt.ylabel("Frequency")
plt.legend()
plt.grid(axis="y", alpha=0.3)

plt.show()

# Calculate standard error
sample_std = data.std()

standard_error = sample_std / np.sqrt(30)

print("\nStandard Error:")
print(round(standard_error, 4))

# Explain Central Limit Theorem
print("\nCentral Limit Theorem:")
print(
    "The Central Limit Theorem states that when random samples "
    "are repeatedly taken, the distribution of their sample means "
    "becomes approximately normal as the sample size increases."
)

# Estimate population mean using sample mean
estimated_population_mean = sample_30_mean

print("\nEstimated Population Mean:")
print(round(estimated_population_mean, 2))

# Final conclusion
print("\nConclusion:")
print(
    "The Anti-Ragging Awareness survey used 75 respondents. "
    "Random samples of 30 observations were taken 100 times "
    "and their means were calculated. The sampling distribution "
    "was concentrated around the original mean. Larger sample "
    "sizes produced more stable sample means. This demonstrates "
    "the Central Limit Theorem and shows that a sample mean can "
    "be used to estimate the population mean."
)
```
